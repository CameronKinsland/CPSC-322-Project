## Project Proposal: Predicting Student Successes and Drop Outs

This is our project's [dataset](https://archive.ics.uci.edu/dataset/697/predict+students+dropout+and+academic+success). The data is contained in a single csv and has 36 features and over 4000 instances. All categorical features are encoded nominally and the feature dictionary is on located on the website. 

The features come from students including whether or not the student is enrolled, graduated, or dropped out. Features include nationality, father's qualification, tuition fee, age at enrollment, scholarship, and country's unemployment rate.

Our class attribute will be "Target", which we'll rename to "status".

Some anticipated challenges is that we have an unequal class distribution with the amount of dropouts being 1421, graduated being 2209, and enrolled being 794. Most of the preprocessing has already been done, but we'll still have to retrace steps and see if there is anything left to do. The creators' emails are listed on the website, so we'll attempt contact to see if we can get additional documentation. Another challenge would be normalization and interpretation because a lot of features are encoded as numbers.

We have 36 features which is relatively large, so we will use visualization, covariances, and corelation coeficients to explore the significant associative features.

The results help predict student success based on given attributes. What attributes have the greatest effect on student success?  
The results would affect students, parents, educators, college administrators, and even taxpayers. How can we, as a society, educational success?
